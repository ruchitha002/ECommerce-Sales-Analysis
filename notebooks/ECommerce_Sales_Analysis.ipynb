```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA = Path("../data")
OUT = Path("../outputs")
OUT.mkdir(exist_ok=True)

orders = pd.read_csv(DATA/"olist_orders_dataset.csv")
items = pd.read_csv(DATA/"olist_order_items_dataset.csv")
products = pd.read_csv(DATA/"olist_products_dataset.csv")
reviews = pd.read_csv(DATA/"olist_order_reviews_dataset.csv")
translation = pd.read_csv(DATA/"product_category_name_translation.csv")

display(orders.head())
display(items.head())
display(products.head())
display(reviews.head())
```

## 1. Data exploration and cleaning

```python
print("Orders shape:", orders.shape)
print("Items shape:", items.shape)
print("Products shape:", products.shape)
print("Reviews shape:", reviews.shape)

print("\nMissing values:")
for name, df in {"orders":orders, "items":items, "products":products, "reviews":reviews}.items():
    print(name)
    display(df.isna().sum().sort_values(ascending=False).head(10))
```

```python
# Convert date columns
date_cols = [
    "order_purchase_timestamp", "order_approved_at",
    "order_delivered_carrier_date", "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
for c in date_cols:
    orders[c] = pd.to_datetime(orders[c], errors="coerce")

# Keep delivered/available orders for sales analysis
orders_clean = orders[orders["order_status"].eq("delivered")].copy()

# Merge order items -> products -> category translation -> reviews
sales = orders_clean[["order_id","customer_id","order_purchase_timestamp"]].merge(
    items[["order_id","order_item_id","product_id","price","freight_value"]],
    on="order_id", how="inner"
).merge(
    products[["product_id","product_category_name"]],
    on="product_id", how="left"
).merge(
    translation, on="product_category_name", how="left"
).merge(
    reviews[["order_id","review_score"]], on="order_id", how="left"
)

sales["category"] = sales["product_category_name_english"].fillna(
    sales["product_category_name"]
).fillna("Unknown")

sales["revenue"] = sales["price"].fillna(0)
sales["month"] = sales["order_purchase_timestamp"].dt.to_period("M").astype(str)

# Documented cleaning decisions
cleaning_notes = pd.DataFrame({
    "Decision": [
        "Converted order timestamp fields to datetime",
        "Restricted sales analysis to delivered orders",
        "Merged order items with products and English category translation",
        "Used Unknown when a product category could not be translated",
        "Used order item price as revenue for category/month analysis",
        "Kept review_score as missing when no review existed"
    ]
})
display(cleaning_notes)
display(sales.head())
```

## 2. Sales Analysis — five questions

```python
# Q1: highest-revenue category
category_revenue = sales.groupby("category")["revenue"].sum().sort_values(ascending=False)
print("Q1 — Highest revenue category:", category_revenue.index[0])
print("Revenue:", round(category_revenue.iloc[0], 2))

# Q2: peak month
monthly_revenue = sales.groupby("month")["revenue"].sum().sort_index()
print("Q2 — Peak sales month:", monthly_revenue.idxmax())
print("Revenue:", round(monthly_revenue.max(), 2))

# Q3: best region
# Olist orders do not contain region directly, so use customer location via customers dataset.
customers = pd.read_csv(DATA/"olist_customers_dataset.csv")
regional = orders_clean[["order_id","customer_id"]].merge(
    customers[["customer_id","customer_state"]], on="customer_id", how="left"
).merge(items[["order_id","price"]], on="order_id", how="left")
state_revenue = regional.groupby("customer_state")["price"].sum().sort_values(ascending=False)
print("Q3 — Best-performing state/region proxy:", state_revenue.index[0])
print("Revenue:", round(state_revenue.iloc[0], 2))

# Q4: average order value trend
order_values = sales.groupby("order_id")["revenue"].sum()
aov = order_values.mean()
monthly_aov = sales.groupby("month").agg(revenue=("revenue","sum"), orders=("order_id","nunique"))
monthly_aov["aov"] = monthly_aov["revenue"] / monthly_aov["orders"]
print("Q4 — Overall average order value:", round(aov, 2))
display(monthly_aov.head())

# Q5: review score distribution
review_distribution = sales.drop_duplicates("order_id")["review_score"].value_counts().sort_index()
display(review_distribution)
```
